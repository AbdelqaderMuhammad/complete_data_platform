```python
import duckdb
```

```python
src_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2024-01.parquet"
```

```python
duckdb.execute(f"""
    COPY (
        SELECT * 
        FROM read_parquet('{src_url}') 
        LIMIT 10000
    ) 
    TO 'yellow_tripdata_sample.parquet' (FORMAT 'PARQUET');
""")

print("Done! File saved as 'yellow_tripdata_sample.parquet'")
```

```python
con = duckdb.execute(f""" 
DESCRIBE SELECT * FROM 'yellow_tripdata_sample.parquet';
""")
```
